# Part 3 — Client-Server Communication

## Server flow
```text
socket() → bind() → listen() → accept() → recv() → send() → close()
```

## Client flow
```text
socket() → connect() → send() → recv() → close()
```

Start the server before the client.

# Part 4 — Create a Basic TCP Server
Create a file called `server.py` and copy this code into it.

````python
import socket

HOST = "127.0.0.1"
PORT = 12345

server_socket = socket.socket(
    socket.AF_INET,
    socket.SOCK_STREAM
)

server_socket.bind((HOST, PORT))
server_socket.listen(1)

print("Server is waiting for a connection...")

client_socket, client_address = server_socket.accept()

print("Connected by:", client_address)

data = client_socket.recv(1024)

message = data.decode()

print("Client says:", message)

reply = "Message received"

client_socket.send(reply.encode())

client_socket.close()
server_socket.close()
````

## Run the Server

# Part 5 — Create a Basic TCP Client
Keep the server terminal open. Create `client.py` and add:

````python

import socket

HOST = "127.0.0.1"
PORT = 12345

client_socket = socket.socket(
    socket.AF_INET,
    socket.SOCK_STREAM
)

client_socket.connect((HOST, PORT))

message = "Hello Server"

client_socket.send(message.encode())

response = client_socket.recv(1024)

print("Server replied:", response.decode())

client_socket.close()
````

# Part 7 — Connect to a Public Web Server
After your localhost client/server works, create `http_client.py`. This connects to `example.com` on port 80 and sends a basic HTTP GET request.

````python
import socket

client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
client.connect(("example.com", 80))

request = (
    "GET / HTTP/1.1\r\n"
    "Host: example.com\r\n"
    "Connection: close\r\n"
    "\r\n"
)

client.send(request.encode())

response = b""
while True:
    data = client.recv(4096)
    if not data:
        break
    response += data

print(response.decode(errors="replace"))
client.close()
````

# Troubleshooting
### `python` command not found
Try `python3 --version` or, on Windows, `py --version`.

### `ConnectionRefusedError`
Make sure the server is running first and both programs use the same host and port.

### `Address already in use`
Stop the old server with `Ctrl + C`, or change the port in **both** files.

### String/bytes error
Remember to use `.encode()` before sending text and `.decode()` after receiving bytes.

### Server seems frozen
If it says `Server is waiting for a connection...`, it is working correctly and waiting for the client.
